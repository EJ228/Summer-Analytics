### 1. What is name of car that has the highest horsepower?  

```python
cars = pd.read_csv("cars.csv")
cars.reset_index(inplace=True)
mhp = cars.loc[cars['horsepower'].idxmax()]
print("Car with highest horsepower:" , mhp['name'])
```

### 2. How many cars have mpg ≥ 35?

```python
count = (cars['mpg']>= 35).sum()
print("Number of cars with mpg greater than equal to 35 is" , count)
```

### 3. What is the most common origin for cars with horsepower > 100 and weight < 3000?

```python
filtcars = cars[(cars['horsepower']>100)& (cars['weight']<3000)]
mco = filtcars['origin'].value_counts().idxmax
print("Most common origin:" , mco)
```

### 4. What is the mean acceleration of cars from Japan? (rounded to 2 decimals)

```python
japcars = cars[cars['origin'] == 'japan']
meanacc = round(japcars['acceleration'].mean(), 2)
print(" Mean acceleration is ", meanacc)
```

### 5. Which year had the highest average mpg?

```python
apby = cars.groupby('model_year')['mpg'].mean()
yearhigh = apby.idxmax()
hampg = round(apby.max(), 2)
print(f" year is: {yearhigh} (Average MPG: {hampg})")
```
